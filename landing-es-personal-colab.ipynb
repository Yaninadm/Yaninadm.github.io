# Landing page — Es Personal by Yai
Este notebook abre la landing en Google Colab, con el catálogo de 39 diseños ordenado por categorías. Ejecutá las celdas en orden y, cuando se solicite, subí el archivo `es-personal-by-yai.zip`.

```python
from google.colab import files
import io, os, zipfile, pathlib

print('Elegí el archivo es-personal-by-yai.zip')
uploaded = files.upload()
zip_name = next(name for name in uploaded if name.lower().endswith('.zip'))
project_dir = pathlib.Path('/content/es-personal-by-yai')
project_dir.mkdir(parents=True, exist_ok=True)
with zipfile.ZipFile(io.BytesIO(uploaded[zip_name])) as archive:
    archive.extractall(project_dir)
print('Proyecto preparado en:', project_dir)
```

```python
import subprocess, time
from google.colab.output import eval_js
from IPython.display import display, HTML

if 'server' in globals() and server.poll() is None:
    server.terminate()
web_dir = project_dir / 'dist'
server = subprocess.Popen(
    ['python3', '-m', 'http.server', '8000', '--directory', str(web_dir)],
    stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL
)
time.sleep(2)
url = eval_js('google.colab.kernel.proxyPort(8000)')
display(HTML(f'<a href="{url}" target="_blank" style="display:inline-block;background:#ff1493;color:white;padding:14px 24px;text-decoration:none;font-weight:bold;border-radius:6px">ABRIR LANDING PAGE</a>'))
```
